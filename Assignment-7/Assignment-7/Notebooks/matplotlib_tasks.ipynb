```python
import matplotlib.pyplot as plt

name = ["Mansur", "Eskendir", "Christina", "Erasyl", "Alua"]
score = [88, 90, 92, 85, 90]
```

```python
# Task 6 

# Линейный график
plt.plot(name, score)
plt.xlabel("Students")
plt.ylabel("SCORE")
plt.title("Student Scores")
plt.show()
```

```python
# Task 7

# Гистограмма
plt.bar(name, score)
plt.xlabel("STUDENT")
plt.ylabel("SCORE")
plt.title("Student Scores")
plt.show()
```
