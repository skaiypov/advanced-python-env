```python
import pandas as pd

# Task 3
data = {
    "NAME": ["Mansur", "Eskendir", "Christina", "Erasyl", "Alua"],
    "AGE": [19, 18, 22, 19, 19],
    "SCORE": [88, 90, 92, 85, 90]
}

df = pd.DataFrame(data)
df
```

```python
# Task 4

print(df.head(3), '\n') # первые 3 строки

print(df.describe().round(2), '\n') # статистика всей таблицы

print("Average score:", df["SCORE"].mean()) # среднее значение оценок
```

```python
# Task 5
print(df[df["AGE"] > 20], '\n') # студенты старше 20 лет

print(df[df["SCORE"] > 80]) # студенты с оценками выше 80
```
